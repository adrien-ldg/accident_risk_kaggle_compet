```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
caracteristiques = pd.DataFrame()
lieux = pd.DataFrame()
usagers = pd.DataFrame()
vehicules = pd.DataFrame()

for i in range(2010, 2012):
    c = pd.read_csv(f"TRAIN/BAAC-Annee-{i}/caracteristiques_{i}.csv", encoding='ISO-8859-1')
    print(c.shape)
    l = pd.read_csv(f"TRAIN/BAAC-Annee-{i}/lieux_{i}.csv", encoding='ISO-8859-1', dtype={'voie': str})
    print(l.shape)
    u = pd.read_csv(f"TRAIN/BAAC-Annee-{i}/usagers_{i}.csv", encoding='ISO-8859-1')
    print(u.shape)
    v = pd.read_csv(f"TRAIN/BAAC-Annee-{i}/vehicules_{i}.csv", encoding='ISO-8859-1')
    print(v.shape)
    caracteristiques = pd.concat([caracteristiques, c], ignore_index=True)
    lieux = pd.concat([lieux, l], ignore_index=True)
    usagers = pd.concat([usagers, u], ignore_index=True)
    vehicules = pd.concat([vehicules, v], ignore_index=True)
    
print("Taille de caracteristiques:", caracteristiques.shape)
print("Taille de lieux:", lieux.shape)
print("Taille de usagers:", usagers.shape)
print("Taille de vehicules:", vehicules.shape)
```
